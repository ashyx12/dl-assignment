```python
import gymnasium as gym
import minigrid

print('Gymnasium:', gym.__version__)
print('MiniGrid:', minigrid.__version__)
```

```python
from minigrid.wrappers import RGBImgPartialObsWrapper

env = gym.make('MiniGrid-Empty-5x5-v0')
env = RGBImgPartialObsWrapper(env)

obs, info = env.reset(seed=42)
print('Observation keys:', obs.keys())
print('RGB shape:', obs['image'].shape)
print('RGB dtype:', obs['image'].dtype)
```

```python
for step in range(10):
    obs, reward, terminated, truncated, info = env.step(env.action_space.sample())
    print(f'step={step:02d} action_space={env.action_space} reward={reward} terminated={terminated} truncated={truncated}')
    if terminated or truncated:
        break

env.close()
```
